```python
import math
def haversine(lat1,lon1,lat2,lon2):
    #convert decimal degrees into radians
    lat1,lon1,lat2,lon2 =map(math.radians,[lat1,lon1,lat2,lon2])
    
    #haversine formula
    dlon= lon2 -lon1
    dlat= lat2-lat1
    a= math.sin(dlat/2)**2 + math.cos(lat1) * math.cos(lat2) * math.sin(dlon/2)**2
    c = 2 * math.asin(math.sqrt(a))
    
   # Radius of earth in kilometers is 6371
    km = 6371 * c
   
    m = km*1000 
    return km,m
```

```python
# Distance between fist point and stationary point
distance_km, distance_m = haversine(-1.3621409, 36.6907849,-1.3630421, 36.6930803)
print("Distance between the two points is {:.2f} kilometers or {:.2f} meters".format(distance_km, distance_m))
```

```python
# Distance between two points
distance_km, distance_m = haversine(-1.3661375, 36.6952998,-1.3630421, 36.6930803)
print("Distance between the two points is {:.2f} kilometers or {:.2f} meters".format(distance_km, distance_m))
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
distances = [81.21, 120.83, 182.29, 225.97, 274.14]  # Distance from stationary point in meters
snr_values = [-19.14, -51.96, -21.65, -25.11, -64]
# Convert 'Out of range' to a value that won't be plotted
snr_values = [float(x) if x != 'Out of range' else np.nan for x in snr_values]

# Plot
plt.figure(figsize=(12, 8))
plt.plot(distances, snr_values, color='gray', linestyle='-', alpha=0.5)
plt.scatter(distances, snr_values, color='blue', marker='x')
plt.title('SNR vs Distance from Stationary Point')
plt.xlabel('Distance from Stationary Point (m)')
plt.ylabel('SNR (dB)')
plt.grid(True)
plt.ylim(-70, 0) 
plt.xticks(distances)
plt.show()
```
